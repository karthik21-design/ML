```python
# ==========================================================
# Experiment No. 2
# Title: Program to Compute Descriptive Statistics
# Dataset: Biostats (biostats.csv)
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_csv("biostats.csv")

print("="*60)
print("FIRST 5 ROWS")
print("="*60)
print(df.head())

print("\n" + "="*60)
print("DATASET INFORMATION")
print("="*60)
df.info()

print("\n" + "="*60)
print("SHAPE OF DATASET")
print("="*60)
print(df.shape)

print("\n" + "="*60)
print("COLUMN NAMES")
print("="*60)
print(df.columns.tolist())

print("\n" + "="*60)
print("SUMMARY STATISTICS")
print("="*60)
print(df.describe())

# -----------------------------
# Select a Numerical Column
# -----------------------------
numeric_columns = df.select_dtypes(include=np.number).columns

print("\nAvailable Numerical Columns:")
print(numeric_columns.tolist())

# Use the first numerical column automatically
column = numeric_columns[0]
data = df[column].dropna()

print(f"\nSelected Column: {column}")

# -----------------------------
# Central Tendency
# -----------------------------
mean = data.mean()
median = data.median()
mode = data.mode()

print("\n========== CENTRAL TENDENCY ==========")
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode.tolist())

# -----------------------------
# Dispersion
# -----------------------------
variance = data.var()
std = data.std()

Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

print("\n========== DISPERSION ==========")
print("Variance:", variance)
print("Standard Deviation:", std)
print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

# -----------------------------
# Outlier Detection
# -----------------------------
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("\n========== OUTLIER DETECTION ==========")
print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Number of Outliers:", len(outliers))

# -----------------------------
# Remove Outliers
# -----------------------------
clean_data = data[(data >= lower) & (data <= upper)]

print("\n========== AFTER REMOVING OUTLIERS ==========")
print("Mean:", clean_data.mean())
print("Median:", clean_data.median())
print("Standard Deviation:", clean_data.std())

# Save cleaned dataset
clean_df = df[df[column].isin(clean_data)]
clean_df.to_csv("cleaned_biostats.csv", index=False)

print("\nCleaned dataset saved as cleaned_biostats.csv")

# -----------------------------
# Histogram
# -----------------------------
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(data, bins=10, color="skyblue", edgecolor="black")
plt.axvline(mean, color="red", linestyle="--", linewidth=2, label="Mean")
plt.axvline(median, color="green", linewidth=2, label="Median")
plt.title(f"Histogram of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.legend()

# -----------------------------
# Boxplot
# -----------------------------
plt.subplot(1,2,2)
plt.boxplot(data, vert=False)
plt.title(f"Boxplot of {column}")

plt.tight_layout()
plt.show()

print("\nExperiment Completed Successfully!")
```
