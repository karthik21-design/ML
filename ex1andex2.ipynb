```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data = {
    "Patient_ID": [101, 102, 103],
    "Name": ["Rahul", "Anita", "Kiran"],
    "Age": [45, 30, 55],
    "Gender": ["Male", "Female", "Male"],
    "Height": [170, 160, 168],
    "Weight": [72, 58, 80],
    "Blood_Pressure": [120, 110, 140],
    "Glucose_Level": [150, 95, 180],
    "Blood_Group": ["O+", "A+", "B+"],
    "Diabetes": ["Yes", "No", "Yes"]
}

# Step 1: Create DataFrame
df = pd.DataFrame(data)

# Display first five rows
print("Dataset:")
print(df.head())

# Step 2: Identify data types
print("\nData Types:")
print(df.dtypes)

# Step 3: Separate numerical and categorical attributes
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
categorical_columns = df.select_dtypes(include=['object']).columns

print("\nNumerical Columns:")
print(list(numerical_columns))

print("\nCategorical Columns:")
print(list(categorical_columns))

# Step 4: Convert categorical variables into numerical form
encoder = LabelEncoder()

for column in categorical_columns:
    df[column] = encoder.fit_transform(df[column])

print("\nDataset after Encoding:")
print(df.head())

# Step 5: Remove unnecessary columns
df = df.drop(columns=['Patient_ID', 'Name'])

print("\nDataset after Removing Columns:")
print(df.head())

# Step 6: Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Step 7: Save cleaned dataset
df.to_csv("hospital_data_cleaned.csv", index=False)

print("\nCleaned dataset saved as hospital_data_cleaned.csv")
```

```python
import pandas as pd

# Sample Dataset
data = {
    "Customer_ID": [101,102,103,104,105,106,107,108],
    "Name": ["Rahul","Anita","Kiran","Sneha","Arjun","Priya","Ravi","Neha"],
    "Age": [25,None,35,40,None,29,32,28],
    "Gender": ["Male","Female",None,"Female","Male","Female",None,"Female"],
    "City": ["Hyderabad","Vizag","Vijayawada",None,"Chennai","Bangalore","Vizag",None],
    "Income": [50000,60000,None,70000,65000,None,72000,58000]
}

df = pd.DataFrame(data)

# Display Original Dataset
print("Original Dataset")
print(df)

# Step 1: Detect Missing Values
print("\nMissing Values")
print(df.isnull())

# Step 2: Count Missing Values
print("\nMissing Values Count")
print(df.isnull().sum())

# Step 3: Replace Numerical Missing Values with Mean

numerical_columns = df.select_dtypes(include=['float64','int64']).columns

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].mean())

# Step 4: Replace Categorical Missing Values with Mode

categorical_columns = df.select_dtypes(include=['object']).columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Step 5: Display Cleaned Dataset

print("\nDataset After Handling Missing Values")
print(df)

# Save Dataset

df.to_csv("customer_cleaned.csv",index=False)

print("\nCleaned dataset saved successfully.")
```
